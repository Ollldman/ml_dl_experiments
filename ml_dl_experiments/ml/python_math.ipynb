# Построение числовых прямых

```python
import matplotlib.pyplot as plt

# Параметры
x_min, x_max = -3, 3          # границы числовой прямой
points = [0]                  # точки, которые нужно отметить
eps = 1.0                     # радиус окрестности (если нужен)
show_epsilon = True           # показывать ли ε-окрестность

# Создаём график
fig, ax = plt.subplots(figsize=(8, 1.2))

# --- Числовая прямая (ось) ---
ax.hlines(0, x_min, x_max, colors='black', linewidth=1.5)  # горизонтальная линия

# Стрелка на конце
arrow_x = x_max
ax.annotate('', xy=(x_max, 0), xytext=(x_max - 0.3, 0),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5))
# Подпись \mathbb{R} рядом со стрелкой (чуть выше или на уровне)
ax.text(arrow_x, 0.01, r'$\mathbb{R}$', fontsize=12, ha='left', va='bottom')

# --- Отмечаем точки ---
for p in points:
    ax.plot(p, 0, 'ko', markersize=8, zorder=5)  # чёрный закрашенный маркер или wo - выколотая
    ax.text(p, 0.08, f'{p}', ha='center', va='bottom', fontsize=12)

# --- ε-окрестность вокруг первой точки (если нужно) ---
if show_epsilon and points:
    a = points[0]
    left, right = a - eps, a + eps
    # Линия окрестности
    ax.hlines(0, left, right, colors='blue', linewidth=3, zorder=4)
    # Концевые точки окрестности
    ax.plot([left, right], [0, 0], 'bo', markersize=6)
    # Подписи под концами
    ax.text(left, -0.08, r'$a-\varepsilon$', ha='center', va='top', fontsize=10, color='blue')
    ax.text(right, -0.08, r'$a+\varepsilon$', ha='center', va='top', fontsize=10, color='blue')

# --- Настройки отображения ---
ax.set_xlim(x_min - 0.2, x_max + 0.2)
ax.set_ylim(-0.2, 0.2)

# Убираем оси и метки
ax.set_xticks([])
ax.set_yticks([])
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

# Сохраняем
plt.tight_layout()
plt.savefig('number_line.png', dpi=200, bbox_inches='tight')
plt.show()
```

# График

```python
import matplotlib.pyplot as plt

# === Настройки ===
# Список точек: [(x1, y1, 'метка1'), (x2, y2, 'метка2'), ...]
points = [
    (2, 3, 'A'),
    (-1, 2, 'B'),
    (0, -2, 'C'),
    # Можно продолжать
]

# Границы графика (автоподстройка возможна, но явное задание надёжнее)
x_min, x_max = -4, 4
y_min, y_max = -3, 4

# Показывать ли подпись ℝ² у стрелки?
show_R2 = True

# === Создание графика ===
fig, ax = plt.subplots(figsize=(7, 6))

# --- Ось X ---
ax.axhline(0, color='black', linewidth=1.5)  # линия y=0
# Стрелка на конце оси X
ax.annotate('', xy=(x_max, 0), xytext=(x_max - 0.3, 0),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5))
ax.text(x_max + 0.1, -0.15, r'$x$', fontsize=14, ha='left', va='top')

# --- Ось Y ---
ax.axvline(0, color='black', linewidth=1.5)  # линия x=0
# Стрелка на конце оси Y
ax.annotate('', xy=(0, y_max), xytext=(0, y_max - 0.3),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5))
ax.text(0.15, y_max + 0.1, r'$y$', fontsize=14, ha='left', va='bottom')

# --- Подпись ℝ² (опционально) ---
if show_R2:
    ax.text(x_max + 0.1, y_max + 0.1, r'$\mathbb{R}^2$', fontsize=14, ha='left', va='bottom')

# --- Отметка точек ---
for (x, y, label) in points:
    ax.plot(x, y, 'ro', markersize=6, zorder=5)  # красная точка
    # Подпись рядом с точкой
    ax.text(x + 0.15, y + 0.15, f'{label}$({x}, {y})$', fontsize=12, ha='left', va='bottom')

# --- Настройка пределов и сетки ---
ax.set_xlim(x_min, x_max + 0.8)   # + место для подписей
ax.set_ylim(y_min, y_max + 0.8)

# Включить сетку (опционально)
ax.grid(True, which='both', linestyle='--', linewidth=0.5, alpha=0.7)
ax.set_xticks(range(x_min, x_max+1))
ax.set_yticks(range(y_min, y_max+1))

# Убрать рамку вокруг графика
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
# Оставить только оси через центр (уже нарисованы вручную)

# --- Сохранение ---
plt.tight_layout()
plt.savefig('cartesian_plane.png', dpi=200, bbox_inches='tight')
plt.show()
```
